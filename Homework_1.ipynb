```python
initial_median = df['horsepower'].median()
print("Initial median:", initial_median)
```

```python
most_frequent = df['horsepower'].mode()[0]
print("Most frequent value:", most_frequent)
```

```python
df['horsepower'] = df['horsepower'].fillna(most_frequent)
```

```python
new_median = df['horsepower'].median()
print("New median:", new_median)
```

```python
X = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].iloc[:7].values
```

```python
XTX = X.T.dot(X)
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv.dot(X.T).dot(y)
```

```python
print("Sum of elements:", w.sum())
```
